```python
# Task 1: Sentiment Analysis
# Understands if text is positive or negative
from transformers import pipeline

print("Loading sentiment analysis model...")
sentiment = pipeline("sentiment-analysis")
print("Model loaded! ✅")

# Test with 5 IT incident related examples
texts = [
    "The server is running perfectly with no issues",
    "Critical system failure causing complete outage",
    "Memory usage is high but system is stable",
    "All services restored successfully after maintenance",
    "Network connectivity lost on production servers"
]

print("\nSentiment Analysis Results:")
print("-" * 50)
for text in texts:
    result = sentiment(text)[0]
    print(f"\nText: {text}")
    print(f"Sentiment: {result['label']}")
    print(f"Confidence: {round(result['score'] * 100, 1)}%")
```

```python
# Task 2: Zero-shot Classification
# Classifies text into categories WITHOUT training
from transformers import pipeline

print("Loading classification model...")
classifier = pipeline("zero-shot-classification")
print("Model loaded! ✅")

# Classify IT incidents into categories
incidents = [
    "Server memory usage reached 95% causing slowdown",
    "Unauthorized login attempt detected on admin server",
    "Database connection failed for 200 users",
    "SSL certificate expiring in 7 days",
    "Batch job failed due to insufficient disk space"
]

# Categories to classify into
categories = ["hardware", "security", "network", "database", "storage"]

print("\nIncident Classification Results:")
print("-" * 50)
for incident in incidents:
    result = classifier(incident, categories)
    top_category = result['labels'][0]
    top_score = round(result['scores'][0] * 100, 1)
    print(f"\nIncident: {incident[:50]}...")
    print(f"Category: {top_category} ({top_score}% confidence)")
```

```python
# Task 3: Text Summarization using question-answering approach
# Works with your current transformers version
from transformers import pipeline

print("Loading text generation model...")
generator = pipeline("text-generation",
                     model="gpt2")
print("Model loaded! ✅")

# Long incident description
long_incident = """
At 2:30 AM on Monday, the production server PROD-01 experienced
a critical memory spike reaching 95% utilization. This caused
significant slowdown across the application layer affecting
approximately 500 concurrent users. The on-call team was
immediately alerted via ServiceNow. Initial investigation
revealed a memory leak in the payment processing module.
The team executed the self-healing automation script which
cleared the memory cache and restarted the affected services.
Full service restoration was achieved within 45 minutes.
"""

# Generate a continuation/summary
result = generator(
    "Summarize this IT incident: " + long_incident[:200],
    max_new_tokens=50,
    num_return_sequences=1,
    truncation=True
)

print("\nOriginal incident length:",
      len(long_incident.split()), "words")
print("\nGenerated summary:")
print(result[0]['generated_text'][-200:])
```

```python
# Cell 5: Combined Incident Analyser
# Sentiment + Zero-shot Classification
from transformers import pipeline

print("Loading models...")
sentiment_model = pipeline("sentiment-analysis")
classifier_model = pipeline("zero-shot-classification")
print("Models ready! ✅")

categories = ["hardware", "security", "network",
              "database", "storage", "application"]

def analyse_incident(incident_text):
    print("=" * 60)
    print(f"INCIDENT: {incident_text[:80]}...")
    print("=" * 60)

    # Sentiment
    sentiment = sentiment_model(incident_text[:512])[0]
    print(f"\n1. SEVERITY SENTIMENT:")
    print(f"   {sentiment['label']} "
          f"({round(sentiment['score']*100, 1)}% confidence)")

    # Category — top 3
    category = classifier_model(incident_text[:512], categories)
    print(f"\n2. INCIDENT CATEGORY — Top 3:")
    for i in range(3):
        print(f"   {i+1}. {category['labels'][i]} "
              f"({round(category['scores'][i]*100, 1)}%)")

    print("=" * 60)

# Test with 3 incidents
analyse_incident(
    "Production database crashed at midnight causing complete "
    "application downtime for all 500 users. DBA team found "
    "corrupted index files and restored service after 2 hours."
)

analyse_incident(
    "Multiple failed login attempts detected on admin portal "
    "from unknown IP addresses. Security team blocked IPs and "
    "forced password reset for all admin accounts immediately."
)

analyse_incident(
    "Memory leak detected on application server causing "
    "gradual performance degradation over 6 hours affecting "
    "response times across all production services."
)
```
